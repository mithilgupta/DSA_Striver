# 917. Reverse Only Letters

```python
def reverseOnlyLetters(s):

    s = list(s)

    l = 0
    r = len(s)-1

    while l<r:
        # print(l,r)
        
        if s[l].isalpha() == False:
            l+=1

        if s[r].isalpha() == False:
            r -=1


        if s[r].isalpha() and s[l].isalpha():
            s[l], s[r] = s[r], s[l]

            l+=1
            r -=1

        # print(s)

        

    return ''.join(s)

    
    
```

```python
reverseOnlyLetters("ab-cd")
```

# 27. Remove Element

```python
def removeElement(nums, val):

    if len(nums)< 2:
        return nums

    l = len(nums)-1

    for r in range(len(nums)-1,-1,-1):
        # print(nums, nums[r], l)

        if nums[r] == val:
            nums[l], nums[r] = nums[r], nums[l]
            l-=1


    return nums, l+1
        

    
    
    
```

```python
removeElement([1], 1)
```

# 977. Squares of a Sorted Array

```python
def sortedSquares(nums):


    output = [0.0]* len(nums)


    l = 0
    r = len(nums)-1

    output_r = len(nums)-1

    while l <= r:

        if nums[l]*nums[l] > nums[r]*nums[r]:
            output[output_r] = nums[l]*nums[l]
            output_r -=1
            l+=1

        else:
            output[output_r] = nums[r]*nums[r]
            output_r -=1
            r-=1

    return output

        
    
    
```

```python
sortedSquares([-4,-1,1,3,10])
```

# 905. Sort Array By Parity

```python
def sortArrayByParity(nums):

    if len(nums)< 2:
        return nums

    l = len(nums)-1

    for r in range(len(nums)-1,-1,-1):
        # print(nums, nums[r], l)

        if nums[r] %2 != 0:
            nums[l], nums[r] = nums[r], nums[l]
            l-=1


    return nums
```

```python
sortArrayByParity([3,1,2,4])
```

# Reverse Prefix

```python
def reversePrefix(word,ch):

    word = list(word)

    l = 0

    r = float('inf')

    for j in range(len(word)):
        if word[j] == ch:
            r = j
            break

    if r == float('inf'):
        return word
    
    while l<r:
        word[l], word[r] = word[r], word[l]
        l+=1
        r-=1


    return ''.join(word)
    

    
    
```

# 541. Reverse String II

```python
def reverseStr( s, k):

    s = list(s)

    l = 0

    while (l*2*k)<len(s):

        rev_l = l*2*k
        rev_r = min((l*2*k)+k-1, len(s)-1)

        while rev_l < rev_r:

            s[rev_l], s[rev_r] = s[rev_r], s[rev_l]
            rev_l +=1
            rev_r -=1

        l+=1

    return ''.join(s)

    
    
```

```python
reverseStr( 'abcdefg', 2) # 0,2 -> 3,4,5 -> 
```
