```python
def check_sum(nums, targetSum):

    temp_cnt = 0

    for i in range(len(nums)):
        for j in range(i+1,len(nums)+1):
            # print(nums[i:j])
            if sum(nums[i:j]) == targetSum:
                temp_cnt +=1


    return temp_cnt
    
```

```python
check_sum([10,5,3,0], 8)
```
