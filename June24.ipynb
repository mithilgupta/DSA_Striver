# Find Duplicate

```python
def findDuplicate(nums):
    slow = 0
    fast = 0

    while True:
        print(nums[slow], nums[fast])
        slow = (slow+1) % len(nums)
        fast = (fast+2) % len(nums)

        if nums[slow] == nums[fast]:
            return nums[slow]

    
    
```

```python
findDuplicate([18,13,14,17,9,19,7,17,4,6,17,5,11,10,2,15,8,12,16,17])
```

```python
def sumOfSeries(n):

    if n == 1:
        return 1

    else:

        return n**3 + sumOfSeries(n-1)
    
```

# 509. Fibonacci Number

```python
def fib(n):


    if n == 0:
        return 0

    elif n == 1:
        return 1

    else:
        return fib(n-1) + fib(n-2)
    
```

```python
fib(4)
```
